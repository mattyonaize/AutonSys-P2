```python
# imports

import gymnasium as gym
import numpy as np
```

```python
env = gym.make("Taxi-v3")

state, info = env.reset(seed=42)

print(state)
print(env.action_space.n)
print(env.observation_space.n)
```

```python
q_table = np.zeros((env.observation_space.n, env.action_space.n))
```

#### Policy

```python
def epsilon_greedy_policy(state, Q, epsilon):
    if np.random.uniform(0, 1) < epsilon:
        return np.random.randint(0, Q.shape[1])
    else:
        q_values = Q[state[0], state[1]]
        max_q = np.max(q_values)
        max_actions = np.where(q_values == max_q)[0]
        return np.random.choice(max_actions)
```

```python
def epsilon_greedy(state, q_table, epsilon):
    if np.random.random() < epsilon:
        return env.action_space.sample()
    else:
        return np.argmax(q_table[state])
```

```python
alpha = 0.1
gamma = 0.99
epsilon = 0.1
epsilon_decay = max(0.01, epsilon * 0.99)

episodes = 1000

env = gym.make("Taxi-v3", render_mode='human')

for episode in range(episodes):
    state, info = env.reset()
    action = epsilon_greedy(state, q_table, epsilon_decay)
    done = False

    while not done:
        next_state, reward, terminated, truncated, info = env.step(action)
        done = terminated or truncated

        next_action = epsilon_greedy(next_state, q_table, epsilon_decay)

        prev_value = q_table[state, action]
        next_value = q_table[next_state, next_action]

        q_table[state, action] = prev_value + alpha * (
            reward + gamma * next_value - prev_value
        )

        state = next_state
        action = next_action
```
